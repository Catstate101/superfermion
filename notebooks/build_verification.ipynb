# Superfermion Build Verification

Rust core import check + README quick-start, executed with the `python3` ipykernel.

```python
import sys
print("Python:", sys.version)
import superfermion as sf
print("sf version:", sf.__version__)
print("Rust core State:", sf.State)
```

```python
# README Quick Start: Bell state with shots
qc = sf.Circuit(2).h(0).cx(0, 1)
result = sf.run(qc, device="cpu", shots=1024)
print("counts:", result.counts)
```

```python
# Exact simulation with sf.simulate()
state = sf.simulate(qc, device="cpu")
print("statevector:", state.numpy())
print("entropy:", state.entropy())
print("purity:", state.purity())
```

```python
# Expectation value (Rust-native)
zz_obs = [([3, 3], 1.0, 0.0)]  # ZZ observable
print("ZZ expectation:", state.expectation(zz_obs))
```

```python
# Parameterized circuit with gradient
qc2 = sf.Circuit(1).ry(sf.param("theta"), 0)
bound = qc2.bind({"theta": 0.5})
state2 = sf.simulate(bound, device="cpu")
grads = state2.grad([([3], 1.0, 0.0)], qc2.to_ir(), {"theta": 0.5})
print("grads:", grads)
```
